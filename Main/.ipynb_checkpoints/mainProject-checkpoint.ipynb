And whether it can be responsibly used.

We are exploring the COVID-19 Clinical Trials dataset using the items of Microsoft Datasheets for Datasets framework. We are looking for the dataset’s origins, limits, ethical concerns, and potential biases. This is important because companies will use datasheet items to analyze datasets like these, especially human subjects, often containing hidden issues that can affect the accuracy and fairness of the conclusions made.

## Data Collection Process

<b>How was the data collected? (e.g., hardware apparatus/sensor, manual human curation, software program, software interface/API; how were these constructs/measures/methods validated?)<b>

Data was collected by using manual human entry into a structured web-based software program. The ClinicalTrials.gov went through the Protocol Registration and Results System (PRS). The PRS includes automated validation rules. 

<b>Who was involved in the data collection process? (e.g.,students, crowdworkers) How were they compensated? (e.g., how much were crowdworkers paid?)<b>

The Kaggle dataset is only a snapshot of COVID-19 clinical trials, created by an individual. The original data is funded by the NIH and NLM. No funding was given to the Kaggle individual. 

<b>Over what time-frame was the data collected? Does the collection time-frame match the creation time-frame?<b>

The data for each trial was added and changed over time as the trial progressed. This dataset on Kaggle is just a single copy of the data taken at one specific time-frame. The government data is constantly being updated while the Kaggle file is frozen. 

<b>How was the data associated with each instance acquired? Was the data directly observable (e.g., raw text, movie ratings), reported by subjects (e.g., survey responses), or indirectly inferred/derived from other data (e.g., part of speech tags; model-based guesses for age or language)? If the latter two, were they validated/verified and if so how?<b>

The data is self-reported by research teams and put into the ClinicalTrials system. The database managers only only see if the information provided was filled out correctly. Not if it is true. The data is what the researchers report. It is not confirmed information. 

<b>Does the dataset contain all possible instances? Or is it, for instance, a sample (not necessarily random) from a larger set of instances?<b>

No. This dataset is a sample from a larger datapool.

<b>If the dataset is a sample, then what is the population? What was the sampling strategy (e.g., deterministic, probabilistic with specific sampling probabilities)? Is the sample representative of the larger set (e.g., geographic coverage)? If not, why not (e.g., to cover a more diverse range of instances)? How does this affect possible uses?<b>

The population is all of the COVID-19 related trials that were registered on ClinicalTrials.gov. The sample strategy was using Kaggle to filter keywords to include every trial on the site about COVID in the USA. The sample is not fully representative of all global COVID research. This means the analyses from the data are best to understand the US COVID research, not the whole world. 

<b>Is there information missing from the dataset and why? (this does not include intentionally dropped instances; it might include, e.g., redacted text, withheld documents) Is this data missing because it was unavailable? <b>

Yes, significantly. The Study Results column has an overwhelming amount of "No Results Available." This is the most critical missing information. This is because the trial sponsors did not submit this information to the original government database. Also the creator of the Kaggle dataset looked for words such as “COVID-19” if a study about the virus used a different name, then the search may have missed it. So, some relevant data are absent because it wasn't found by specific keywording.
